```python
# we can also create a sequence of numbers using linspace. The difference between arange and linspace is that in arange we specify the step size, whereas in linspace we specify the number of points we want to generate between the start and end values.

f=np.linspace(1,30,5)
f
```

```python
# create 1d 2d 3d arrays

g=np.arange(10) # 1d array
print(g)

h=np.arange(12).reshape(4,3) # 2d array
print(h)

i=np.arange(6).reshape(3,2,1) # 3d array
print(i)
```
